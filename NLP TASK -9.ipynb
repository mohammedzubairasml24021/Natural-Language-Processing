```python
from collections import Counter
counts = Counter(['hello', 'hello', 'world'])
total = sum(counts.values())
probs = {w: c/total for w,c in counts.items()}

# Explicitly check a word that was not seen
unseen_word = 'apple'
naive_prob = probs.get(unseen_word, 0)   # .get returns 0 if word missing
print("Naïve probs:", probs)
print(f"Probability of unseen word '{unseen_word}':", naive_prob)
```

```python
def laplace_smooth(counts, vocab_size):
    """Return smoothed probabilities and unseen‑word probability."""
    total = sum(counts.values()) + vocab_size
    smoothed = {w: (c+1)/total for w,c in counts.items()}
    unseen = 1/total
    return smoothed, unseen

counts = Counter(['hello', 'hello', 'world'])
vocab_size = 10
probs, unseen = laplace_smooth(counts, vocab_size)
print("Smoothed probs:", probs)
print("Unseen‑word probability (Laplace):", unseen)
```

```python
# Include the unseen word probability for every possible vocab entry
all_probs = list(probs.values()) + [unseen]*(vocab_size - len(counts))
total_sum = sum(all_probs)
print("Sum of all probabilities (incl. unseen):", total_sum)
print("Any zero probability?", any(p==0 for p in all_probs))
```
